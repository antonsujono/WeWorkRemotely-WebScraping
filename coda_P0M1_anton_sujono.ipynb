```python
# data manipulation or dataframe
import pandas as pd

# webscraping
from selenium import webdriver
from bs4 import BeautifulSoup
from time import sleep
```

```python
# Membuat Driver
driver = webdriver.Chrome()

#Link Jobstreet
url = "https://weworkremotely.com/remote-software-developer-jobs"

# Buka halaman
driver.get(url)

# Tunggu halaman terbuka
sleep(3)

#Scroll agar semua produk dimuat
for i in range(30):
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    sleep(2)

# Baru ambil HTML
html = driver.page_source

# Buat BeautifulSoup
soup = BeautifulSoup(html, "html.parser")
```

```python
# 1. Empty List untuk menampung data
jobs = []
companies = []
locations = []
links = []

#Ambil Semua Box
boxes = soup.select("li.new-listing-container")

# Looping untuk ambil data
for box in boxes:
    #Judul Pekerjaan
    try:
        job_item = box.find('span', {'class': 'new-listing__header__title__text'}).get_text()
        jobs.append(job_item)
    except:
        jobs.append(None)

    #Nama Perusahaan
    try:
        company_item = box.find('p', {'class': 'new-listing__company-name'}).get_text()
        companies.append(company_item)
    except:
        companies.append(None)
        
    #Lokasi
    try:
        location_item = box.find('p', {'class': 'new-listing__company-headquarters'}).get_text()
        locations.append(location_item)
    except:
        locations.append(None)

    #Link Detail
    try:
        link_item = box.find('a', {'class': 'listing-link--unlocked'}).get('href')
        links.append(link_item)
    except:
        links.append(None)

driver.quit()
```

```python
# Membuat DataFrame
df = pd.DataFrame({
    'Judul Pekerjaan': jobs,
    'Nama Perusahaan': companies,
    'Lokasi / Region': locations,
    'Link Detail': links
})
```

```python
# Informasi DataFrame
df.info()
```

```python
# Cek data yang memiliki nilai null pada kolom "Lokasi / Region"
df[df["Lokasi / Region"].isnull()]
```

```python
# Mengisi nilai kosong pada kolom "Lokasi / Region" dengan "Unknown"
df["Lokasi / Region"] = df["Lokasi / Region"].fillna("Unknown")
```

```python
# Hapus data duplikat karena ada beberapa data yang sama persis
df = df.drop_duplicates()
```
